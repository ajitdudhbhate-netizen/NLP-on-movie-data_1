```python
# using for loop to get the genre of each movie
# and dict to store the genre and its count
```

```python
import ast
```

```python
def genre_extract(col):
    l=[]
    for i in ast.literal_eval(col):
        l.append(i['name'])
    return l    
```

```python
data['genres']=data['genres'].apply(genre_extract)
```

```python
def keywords_extract(col):
    l=[]
    for i in ast.literal_eval(col):
        l.append(i['name'])
    return l    
```

```python
data['keywords']=data['keywords'].apply(keywords_extract)
```
